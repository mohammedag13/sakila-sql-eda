## EDA: Exploratory Data Analysis

## film table
1. How many films are in the database?

```python
%%sql
select
    count(film_id) as number_of_films
from film;
```

There are 1000 films in the sakila database

2. what are the shortest and longest film lengths?

```python
%%sql
select
    min(length) as shortest_length,
    max(length) as longest_length
from film;
```

The shortest film is 46 minutes long, and the longest film is 185 minutes long

3. What ratings exist?

```python
%%sql
select 
    distinct rating
from film;
```

The ratings that exist are: 'G','PG','PG-13','R','NC-17'

4. How many films belong to each rating?

```python
%%sql
select 
    rating,
    count(film_id) as number_of_films
from film
group by rating;
```

Number of films that belong to each rating:
G	178
PG	194
PG-13	223
R	195
NC-17	210

5. How many films belong to each category?

```python
%%sql
select c.name as category_name, count(f.film_id) as number_of_films
from film as f 
inner join film_category as fc
on f.film_id = fc.film_id
inner join category as c
on fc.category_id = c.category_id
group by c.name
order by number_of_films;
```

## Which films and categories perform best?

```python
%%sql 
select 
    f.title,
    count(r.rental_id) as number_of_rentals
from film as f
inner join inventory as i
on f.film_id = i.film_id
inner join rental as r 
on i.inventory_id = r.inventory_id
group by f.title
order by number_of_rentals desc;
```

### Key Takeaway: Film Rental Distribution
Analysis of individual film rentals reveals a well-distributed demand across the catalogue. There are no extreme outliers or "blockbuster" titles that dominate the rental volume. Instead, customer interest is spread fairly evenly across a wide variety of films.


## Which categories that perform the best?

```python
%%sql 
select 
    c.name,
    count(rental_id) as number_of_rentals_per_category
from film as f 
inner join film_category as fc
on f.film_id = fc.film_id
inner join category as c
on c.category_id = fc.category_id
inner join inventory as i
on f.film_id = i.film_id
inner join rental as r
on i.inventory_id = r.inventory_id
group by c.name
order by number_of_rentals_per_category desc;
```

### Key Takeaway: Category Performance
* **Balanced Demand:** Rental volume across categories shows a relatively even distribution, ranging from 1,179 rentals (Sports) down to 830 rentals (Music).
* **No Single Dominant Genre:** While top categories like Sports, Animation, and Action perform slightly better, no single category overwhelmingly dominates customer demand.
* **Business Implication:** The store's catalog appeals evenly across genres, suggesting that customer demand is broadly distributed rather than concentrated in specific niche categories.

```python
%%sql
select 
    s.store_id,
    count(r.rental_id) as num_of_rentals
from inventory as i
inner join rental as r
on i.inventory_id = r.inventory_id
inner join store as s 
on s.store_id = i.store_id
group by s.store_id;
```

### Key Takeaway: Store Performance Comparison
* **Balanced Store Volume:** Total rental volume is almost evenly split between both locations, with Store 1 handling 7,923 rentals (49.4%) and Store 2 handling 8,121 rentals (50.6%).
* **Consistent Operations:** Neither store serves as a primary hub or dominant revenue driver in terms of volume; operational demand is balanced across locations.
* **Overall Conclusion:** Across individual films, categories, and physical stores, the Sakila rental business exhibits remarkably uniform demand distribution with no significant outliers.
